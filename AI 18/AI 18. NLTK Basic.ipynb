# Токенизация

```python
from nltk.tokenize import word_tokenize
text = "Hello World! How are you? I'm good at work! Natural language processing and data science is great!"

tokens = word_tokenize(text)

print(f"Text: {text}")
print(f"Tokens: {tokens}")
print(len(tokens))
```

```python
from nltk.tokenize import sent_tokenize
sentences = sent_tokenize(text)

for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence.strip()}")
```

```python
from nltk.tokenize import TreebankWordTokenizer, WordPunctTokenizer, RegexpTokenizer

text = """Please don't stop the music (music, music, music, music)
It's gettin' late, I'm making my way over to my favorite place
I gotta get my body moving, shake the stress away
I wasn't looking for nobody when you looked my way
Possible candidate, yeah
"""
tree = TreebankWordTokenizer()
print(tree.tokenize(text))
print()
word_punct = WordPunctTokenizer()
print(word_punct.tokenize(text))

print()

regexp = RegexpTokenizer(r'\w+')
print(regexp.tokenize(text))
```

# Лемматизация

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["begin", "began", 'begun', 'runner', 'running', 'runs', 'ran', 'better', 'mice', 'cats', 'geese']
for word in words:
    lemma = lemmatizer.lemmatize(word)
    print(f"{word:15}: {lemma}")
```

```python
# pos = 'v' - глагол, pos = 'n' - существительное, pos = 'a' - прилагательное, pos = 'r' - наречие
for word in words:
    print(f"{word:15} - verb: {lemmatizer.lemmatize(word, pos='v')}")
    print(f"{word:15} - adjective: {lemmatizer.lemmatize(word, pos='a')}")
    print(f"{word:15} - noun: {lemmatizer.lemmatize(word, pos='n')}")
    print()

```

```python
from nltk import pos_tag
from nltk.corpus import wordnet

def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

sentence = "The children were running quickly through the old garden, while their teacher, who had written several books, was carefully looking after the mice that had gone missing before they finally gave up searching."

tokens = word_tokenize(sentence)
pos_tags = pos_tag(tokens)
print(pos_tags)
print()

for word, pos in pos_tags:
    wordnet_pos = get_wordnet_pos(pos)
    lemma = lemmatizer.lemmatize(word.lower(), pos=wordnet_pos)
    print(f"{word:15}: {pos:3}-> {lemma}")
```

# Стемминг

```python
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer

porter = PorterStemmer()
stemmer = SnowballStemmer("english")
lancaster = LancasterStemmer()

words = words = [
    "running",
    "better",
    "children",
    "unbelievable",
    "mice",
    "written",
    "studying",
    "quickly",
    "geese",
    "improved",
    "overthinking",
    "berries",
    "driving",
    "happiest",
    "crises",
    "redoing",
    "groundbreaking",
    "wolves",
    "categorization",
    "backup"
]

for word in words:
    p = porter.stem(word)
    s = stemmer.stem(word)
    l = lancaster.stem(word)
    print(f"{word:15}: {p}-> {s}-> {l}")

```

```python
from nltk.corpus import brown, gutenberg
print(brown.categories())
print()
news_words = brown.words(categories='news')
print(list(news_words[:30]))
print(len(news_words))

news_sents = brown.sents(categories='news')
print(len(news_sents))
print(news_sents[3])
```

```python
# for file_id in gutenberg.fileids():
#     print(file_id)

alice = gutenberg.words('carroll-alice.txt')
alice_sents = gutenberg.sents('carroll-alice.txt')
print(len(alice))
print(len(alice_sents))
print(alice_sents[5])
```

```python
from nltk import FreqDist

alice_words_lower = [word.lower() for word in alice if word.isalpha()]

freq_dist = FreqDist(alice_words_lower)

for word, count in freq_dist.most_common(15):
    print(f"{word:<15}: {count:>5}")

print()

print(freq_dist.hapaxes()[: 10])
print(len(freq_dist.hapaxes()))
```
